````python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd

df = pd.read_csv(PROJECT_ROOT / "data/sample/tickets.csv")
print(df.shape)
df.head(10)
````

````python
# Distribuicao de classes
df["categoria"].value_counts().plot(kind="bar", title="Distribuicao de classes");
````

````python
# Qualidade dos dados: ausentes e duplicatas (inseridos de proposito pelo gerador)
print("Valores ausentes por coluna:")
print(df.isna().sum())
print("\nDuplicatas:", df.duplicated(subset=["texto", "categoria"]).sum())
````

````python
# Distribuicao do tamanho dos textos
df["texto"].dropna().str.len().plot(kind="hist", bins=30, title="Tamanho dos textos (caracteres)");
````
